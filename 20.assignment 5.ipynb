```python
# Assignment-5 on Loops 

#1: print in the same line

# Print two strings on the same line with '*' between them and ' * ' at the end
print("Hello", "Aditya's", sep="*", end=" * ")

# Since 'end' didn't create a new line, this prints immediately after the previous output
print("zizzag")

# Initialize counter for the loop
i = 1

# Loop runs while i is less than 4 (1, 2, 3)
while i < 4:
    # Print text with current number, stay on same line
    print(f"Hello Aditya {i}", end=" ")
    # Increment counter
    i += 1

# Reset counter for the next loop
i = 1

# Second loop runs 3 times again
while i < 4:
    # Print text with '*' separator and space at the end
    print(f"Hello{i}", "Aditya", sep="*", end=" ")
    # Increment counter
    i += 1
```

```python
#2: star pattern 

#2.a: Triangle Pattern 
# Set total rows for the triangle
n = 5

# Outer loop: Runs 1 to 5, controlling the row number
for i in range(1, n+1):
    # Inner loop: Runs 1 to 'i', printing stars for the current row
    for j in range(1, i+1):
        # Print a star followed by a space, staying on the same line
        print("*", end = " ")
    # Move to the next line after finishing the current row's stars
    print()

# --- Shortcut Method ---

# Outer loop: Runs 1 to 5
for i in range(1, n+1):
    # Print 'i' number of stars in one go using string multiplication
    print("* " * i)
```

```python
#2.b: inverted triangle 

# Set total rows for the inverted triangle
n = 5

# Outer loop: Starts at 5 and counts down to 1
for i in range(n, 0, -1):
    # Inner loop: Prints 'i' stars for the current row
    for j in range(1, i+1):
        # Print a star and stay on the same line
        print("*", end = " ")
    # Move to the next line after finishing the row
    print()

# --- Shortcut Method ---

# Outer loop: Starts at 5 and counts down to 1
for i in range(n, 0, -1):
    # Print 'i' stars at once using string multiplication
    print("* " * i)
```

```python
#2.c: pyramid pattern 

# Set total rows for the pyramid
n = 5

# Outer loop: Runs from 1 to 5 (1st row to last row)
for i in range(1, n+1):
    # Add spaces to push stars to the center: (Total Rows - Current Row)
    print(' ' * (n - i), end="")
    # Add stars: Formula (2 * i - 1) gives 1, 3, 5, 7, 9 stars
    print("*" * (2 * i - 1))

# --- Shortcut Method ---

# Outer loop: Runs from 1 to 5
for i in range(1, n+1):
    # Combine spaces and stars in a single print statement
    # ' ' * (n - i) creates leading spaces
    # "*" * (2 * i - 1) creates the growing star pattern
    print(' ' * (n - i) + "*" * (2 * i - 1))
```

```python
#3: Factorial of a number 

# Define a function named factorial that takes one number (n)
def factorial(n):
    # Initialize result to 1 (since multiplying by 1 doesn't change the value)
    result = 1
    
    # Loop continues as long as n is greater than 0
    while n > 0:
        # Multiply current result by n and store it back in result
        result *= n 
        # Decrease n by 1 to move to the next number in the sequence
        n -= 1 
        
    # Return the final calculated result
    return result 

# Call the function with 5 and print the output
print(factorial(5))
# 5! means 5 * 4 * 3 * 2 * 1 = 120
```

```python
#4: Count vowels in a string 

# Define the text string to analyze
my_string = "Aditya Prakash Acharya"

# Define a string containing all lowercase vowels
vowels = "aeiou"

# Initialize a counter variable to 0
count = 0

# Loop through every single character in the string
for char in my_string:
    # Convert the current character to lowercase and check if it exists in the vowels string
    if char.lower() in vowels:
        # If it is a vowel, increase the count by 1
        count += 1 

# Print the final total count of vowels found
print("Number of vowels are", count)
```

```python
#5: Longest word in a string 

# Define the sentence to analyze
sentence = "Aditya Prakash Acharya"

# Split the sentence into a list of individual words (splits by spaces)
words = sentence.split()

# Initialize a variable to store the longest word found so far (starts empty)
longest_word = ""

# Loop through each word in the list
for word in words:
    # Check if the current word's length is greater than the longest word found so far
    if len(word) > len(longest_word):
        # If yes, update 'longest_word' to this new word
        longest_word = word 

# Print the final result
print("The longest word is:", longest_word)
```

```python
#6: do-while loop in python 

# Start an infinite loop that keeps running until we explicitly break it
while True:
    # Ask the user to enter a number and convert it to an integer
    num = int(input("Enter a number greater than 10: "))

    # Check if the entered number is greater than 10
    if num > 10:
        # If valid, print a success message
        print(f"Valid number entered: {num}") 
        # Stop the loop immediately using 'break'
        break 
    else:
        # If invalid, tell the user to try again
        print("Number is not greater than 10, try again!")
```

```python
#7: Fibonacci Sequence 

# Define a function to generate the Fibonacci sequence up to n numbers
def fibonacci(n):
    # Initialize the first two numbers: 'a' is 0, 'b' is 1
    a, b = 0, 1
    # Initialize a counter to track how many numbers have been printed
    count = 0
    
    # Loop until we have printed 'n' numbers
    while count < n:
        # Print the current number 'a'
        print(a)
        
        # Update 'a' to the old value of 'b'
        # Update 'b' to the sum of the old 'a' and 'b' (this creates the next number)
        a, b = b, a + b
        
        # Increase the counter by 1
        count += 1

# Call the function to print the first 10 numbers in the sequence
fibonacci(10)
```
