Write a Simple program to determine if the given year is leap year using user input

```python
# leap year condition
# 4 divisible & 100 not divisible
# 400 divisible

#user imput
year = int(input("Enter he year"))

#checking the leap year
if (year % 4 == 0 and year % 100 != 0) or (year % 4 == 0):
    print(f"{year} its a leap year")
else:
    print(f"{year} not a leap year")
```
