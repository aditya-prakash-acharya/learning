```python
# Nested Loop

# Loop inside another loop is a nested loop.
# this means every single time the outer loops run, the inner loops runs its all of its op iteration.


# why to use ?
# handling multi-dimensional data: such as matrices, grids, or list of lists
# complex iterations: opts depends on multple variables and dimentions
# pattern generations:creating opatters  such a graphics and games

#syntax:
    # Outer_loop:
    #   inner_loop:
       # code block to execute - inner loop
    #code block to execute - outer loop
```

```python
# print numbers from 1 to 3 for 3 times 

for i in range(3):
    # print("Outer loop iteration no, ", i)
    for num in range(1,4):
        print(num) 
    print("- - -")

```

```python
# print numbers from 1 to 3 for 3 times using while-for loop : nested loop 
i = 1

while i < 4:
    print("while loop iteration no.", i)
    for j in range(1,4):
        print(j)
    # print("- - -")
    i += 1 
```
