```python
# 3. Slice to get only the middle character(s): For 
# `my_string = "Aditya"`, use slicing to extract the middle 
# character(s).

my_string = "Aditya"
print(my_string[2:4]) # my output

#AI output for any string any characters even or odd

# 1. Define the string to analyze.
# This string has 15 characters (odd length).
my_string = "Adityabatmanfan"

# 2. Calculate the total number of characters in the string.
# len() counts every character.
# Result: 15
length = len(my_string)

# 3. Find the middle index using integer division.
# '//' divides the number and removes the decimal part.
# 15 // 2 = 7 (the index of the exact middle character).
mid = length // 2

# 4. Check if the total length is an even number.
# The '%' modulus operator gives the remainder. If remainder is 0, it's even.
# If remainder is 1, it's odd.
# For "Adityabatmanfan" (15), 15 % 2 is 1 (so it's ODD).
if length % 2 == 0:
    
    # --- CASE A: EVEN LENGTH (e.g., "Aditya" -> 6 chars) ---
    # If length is even, there are TWO middle characters.
    # We need to slice from (mid-1) to (mid+1) to get both.
    # Example "Aditya": mid=3. Slice [2:4] gives "it".
    middle = my_string[mid-1:mid+1]

else:
    
    # --- CASE B: ODD LENGTH (e.g., "Adityabatmanfan" -> 15 chars) ---
    # If length is odd, there is exactly ONE middle character.
    # We just grab the character at the 'mid' index.
    # Example "Adityabatmanfan": mid=7. Index 7 is 'f'.
    middle = my_string[mid]

# 5. Print the result.
# For "Adityabatmanfan", this will print 'f'.
print(middle)
```

```python
# 6. How to Reverse a String Using Python String Methods?

rev_str = "Monday"
print(rev_str[::-1])

# AI exlanation How it works:
# The syntax [::-1] uses Python's slice notation [start:stop:step]:

# Start: (empty) → Defaults to the end of the string (because step is negative).
# Stop: (empty) → Defaults to the beginning of the string.
# Step: -1 → Moves backwards one character at a time.
```

```python
# 7. Write a Python function to check if a string is a palindrome 
# using string methods.

def check_pal(s):
    return s == s[::-1]

# Test it
print(check_pal("radar"))   # True
print(check_pal("hello"))   # False     

# How it works:

# s[::-1] reverses the string.
# == checks if the original matches the reversed one.
# It returns True or False automatically.              
```

```python
# 8. Difference Between `find()` and `index()` in Python?

# find()	Searches for a substring and returns its index number.	Returns -1 (The program continues running).
# index()	Searches for a substring and returns its index number.	Crashes the program with a ValueError (The program stops).

text = "Python is fun"

# --- CASE 1: The word EXISTS in the string ---
print("1. Searching for 'Python' (Exists):")

# Using find()
pos_find = text.find("Python")
print(f"   find() result: {pos_find}") 
# Output: 0 (It works fine)

# Using index()
pos_index = text.index("Python")
print(f"   index() result: {pos_index}")
# Output: 0 (It also works fine)


# --- CASE 2: The word DOES NOT EXIST in the string ---
print("\n2. Searching for 'Java' (Does NOT exist):")

# Using find() -> Safe
# It returns -1 and does NOT crash the program.
pos_find_missing = text.find("Java")
print(f"   find() result: {pos_find_missing}")
# Output: -1

# Using index() -> CRASHES
# If you uncomment the next line, the program will STOP here with an error.
# pos_index_missing = text.index("Java") 
# print(f"   index() result: {pos_index_missing}")

# To prove find() is safe, we can add this line:
print("   Program continues running after find()!")
```

```python
# 9. Efficient String Concatenation method: Why is using `join()` 
# often more efficient than using concatenation in a loop?

#Using join() is more efficient because it calculates the total length of the resulting string once and allocates memory only once. In contrast, concatenation in a loop creates a new string object in memory for every iteration, causing unnecessary copying of data and consuming more time and memory, especially with large lists.

parts = ["1", "2", "3", "4", "5"]
# Python sees the list, calculates total length (5), 

result = "".join(parts) 
print("Result:", result) 
```
