```python
#create function without parameters
def greetings():
    print("welcome to python course")

#call function with parameters
greetings()
```

```python
#create a function to add two numbers
def add2numbers(a,b): #parameters
    result = a + b
    print("the sum is ", result)

#call baove sum function 
add2numbers(5,5)  #arguments

add2numbers(a=8,b=7)
add2numbers(b=1,a=2)
```

```python
#create a function to add three numbers
def add3numbers(a,b,c): #parameters
    result = a + b + c
    print("the sum is ", result)

#call baove sum function 
add3numbers(5,5,9)  #arguments
```

```python
# return statement is used in a function to send back the result to function where called 
# when return is runs it stops and excutes the value

#function wit return statement
def add2numbers(a,b): #parameters
    return a+b
#return a-b 
sum2num = add2numbers(10,1)
print(sum2num)
```

```python
#function to convert celsius to fahrenheit
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit
#call function
temp_f = celsius_to_fahrenheit(25)
print(temp_f)
print("the type with return is ", type(temp_f))
```

```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    print(fahrenheit)
#call function
temp_f = celsius_to_fahrenheit(25)
print("the type with out return is ", type(temp_f))
```
