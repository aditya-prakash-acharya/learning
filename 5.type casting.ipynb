typecasting is something that coverting from one data type to another data type

we need this when we need to perform special operation or when we need to format data in specific way

```python
# casting in python
a = 1
print(type(a))
```

```python
b = "1"
print(a+b) 
#output will be error (int + str) not possible
```

```python
c = int(b)
print(type(c))
print(a+int(b))
print(a+c)
```

```python
# all strings cant be cast into numerical type
name = "aditya"
newname = int(name)
```

```python
# all numerical type can cast into string
mynum = 26
mynum2 = str(mynum)
print(type(mynum2))
```

```python
f1 = 22.56
f2 = int(f1)
print(f2)
print(type(f2))
```

```python
in1 = 26
print(type(float(in1)))
```

```python
#inplicit type casting 
#this generally converts the data automaticall to the actual value to avoid data loss
value_int = 10
value_float = 10.5
valueopt = value_int+value_float
print(type(valueopt))
print(valueopt) #correct value
```

```python
#Explicit Type casting
#also known as type casting  its done manually and done to ensure desired type conversion to avoid unexpected behaviour

#converting String to integer
str_num = "26"
int_num = int(str_num)
print(int_num)
print(type(int_num))
```

```python
#Converting a value to boolean
a0 = bool(0)
print(a0)
print(type(a0))

a1 = bool(1)
print(a1)
print(type(a1))
```
