Operators in python are special smbolys or kewords used to perform operation on operands (variables and values)

Addition Operator '+': a + b

Equal Operator '==': a == b

and Operator 'and': a > 10 and b < 20

```python
#Operators 
#1. Arithmetic 
#2. Comparision
#3. Assignment
#4. Logical
#5. Bitwise
#6. Identity
#7. Membership
```

```python
#1. Arithmetic Operators
a = 5
b = 3
print(a + b) #addition
print(a - b) # subtraction
print(a * b) #multiplication
print(a / b) #division
print(a % b) #modulus (remainder)
print(a // b) #flooring int() of division
print(a ** b) #exponentiation
```

```python
#comparision operators output is boolean t or f
a = 5
b = 3
print(a > b) #greater than operator
print(a < b) #smaller than operator
print(a == b) #equal operator
print(a != b) #not equal operator
print(a <= b) #smaller than equal
print(a >= b) #greaer than equal
```

```python
#3. Assignment operator
a = 5
a = a + 3
a = a - 3
a = a / 3
```

```python
#4. Logical operator

a = 10
b = 20

#and operator rules
#True + True = True
#False + True = False
#False + False = False
print('And Operator')
print(a>10 and b<10)
print(a==10 and b==20)
print(a==10 and b<10)

#or operator rules
#True + false = True
print("Or Opeartor")
print(a>10 and b<10)
print(a==10 and b==20)
print(a==10 and b<10)

#not operator
print("Not Opeartor")
print(not(a==10 and b<10))
```

```python
#5. Identity operator - is,is not
x = [1,2,3]
y = x 
z = [1,2,3]
print(x is y) # is operator
print(x is z)

print(x is not z) #is not operator
```

```python
#6. Membership operators - in, not in
my_list = ['apple', 'orange', 'watermelon']
print('apple' in my_list) #in operator
print('apple2' in my_list) 
print('apple2' not in my_list) #not in operator
```

```python
#7. Bitwise operator
a = 5
b = 3
print(a&b)
```
