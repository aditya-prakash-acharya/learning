```python
# Loops in Python - while loop 
# While loop ececutes block of code as long as given conditions remain true it 
# checks condition eg=]fore each iteration


count = 0 

while count < 5: #condition 
    print(count)
    count = count + 1


# print numbers from 1 to 5 using while loop
count = 1 
while count < 6: #condition 
    print(count)
    # count = count + 1
    count += 1 


count = 5 
while count > 0: #condition 
    print(count)
    # count = count + 1
    count -= 1 
else:
    print("while loop ended")


# while True:
#     print("again and again!!") 
# check conditions to avoid infinite loop
```

```python
# Loops in Python -  for loop 
# for ;oop used ti iterate over a sequence and execute a block of code for each element in sequence

language = 'Python' # sequence 

for x in language:
    print(x)

# range function
# range(stop)
# range(start, stop = isme last tak nahi jayega 0-5 ke count mai 4 tak hi hoga)
# range(start, stop, step)

for i in range(5):  # stop argument
    print(i)

for i in range(5,10):     # start, stop argument
    print(i)

for i in range(1,10,3):  # start, stop, step argument
    print(i)

for i in range(5):
    print(i)
else:
    print("for loop ended")

```

```python
# While Loop VS For Loop

# while loop
# 1. A while lopps keep running as long as conditions is true.
# 2. it is generally used when we dont know how many iterations wil be used and loop continues based on condition.
 
# for loop 
# 1. a loop iterates around sequence and runs the loop of each items in the sequence.
# 2. it is used when we know how many times w want to repat the code.

```

```python

# loop control statements 
# Loop control statements allow you to "alter" = change the normal flow of a loop
# python supports 3 clauses within loops:

# pass statement
# break statement
# continue statement

# 1. pass statement 
# the pass statement  is used as a placeholder(it does nothing) for the future code, and runs the entire code without causing any syntax error.

for i in range(5):
    # code to be updated
    pass 

count = 5 
while count > 0:
    if count == 3:
        pass 
    else:
        print(count)
    count -= 1 
```

```python
#2. break statement 
# The brak statement terminates the loop entirely exiing from it immediately.

for i in range(5):
    if i == 3:
        break 
    print(i) 

print("---------")
```

```python
#3. continue statement 
# the continue statement skips the current iteration and moves to the next one.

for i in range(5):
    if i == 3:
        continue 
    print(i) 
```

```python
# pass statment vs continue statement
count = 5 
while count > 0:
    if count == 3:
        pass 
    else:
        print(count)
    count -= 1 
```

```python
# continue statement: don't try - infinite loop
count = 5 
while count > 0:
    if count == 3:
     continue 
    else:
        print(count)
    count -= 1 
```
