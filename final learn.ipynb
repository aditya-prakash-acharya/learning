```python
# ----------------------------------------------------------------------------
# 1.1 PRINT STATEMENT - How to display output
# ----------------------------------------------------------------------------

print("=" * 60)
print("PART 1: PYTHON BASICS")
print("=" * 60)

# Basic print - displays text
print("hello world")
print("hello world")

# Print with multiple items - sep parameter controls separator
print("hello", "world", "python", sep=" - ")
# Output: hello - world - python

# Print with end parameter - controls what comes at the end (default is newline)
print("Hello", end=" ")
print("World")
# Output: Hello World (on same line)

print("First line", end=" | ")
print("Second line")
# Output: First line | Second line
```

```python
# ----------------------------------------------------------------------------
# 1.3 VARIABLES - Store values for later use
# ----------------------------------------------------------------------------

# Variables are containers for storing data values
a = 5  # integer variable
print(a)

# Variable naming rules:
# 1. Must start with letter or underscore (_)
# 2. Can contain letters, numbers, underscore
# 3. Case sensitive (name, Name, NAME are different)
# 4. Cannot use reserved words (for, if, while, etc.)

my_name = "Aditya"        # snake_case (preferred in Python)
myName = "Aditya"         # camelCase
MyName = "Aditya"         # PascalCase
myname = "Aditya"         # flat_case
_my_name = "Aditya"       # underscore start

print("My name is:", my_name)

# Multiple assignment
x, y, z = 10, 20, 30
print("Multiple assignment:", x, y, z)

# Assign same value to multiple variables
p = q = r = 100
print("Same value:", p, q, r)

# Variables can change type
var = 10       # integer
var = "hello"  # now string
var = 3.14     # now float
print("Variable changing type:", var)
```

```python
# ----------------------------------------------------------------------------
# 1.4 ESCAPE CHARACTERS - Special characters in strings
# ----------------------------------------------------------------------------

print("\n--- Escape Characters ---")
print("Hello\nWorld")        # \n = new line
print("Hello\tWorld")        # \t = tab (spaces)
print("He said, \"Hello!\"")  # \" = double quote inside string
print('It\'s easy')          # \' = single quote inside string
print("Backslash: \\")       # \\ = backslash

# Raw string - ignores escape characters
print(r"Hello\nWorld")       # Prints: Hello\nWorld
```

```python
# ============================================================================
# PART 2: DATA TYPES IN PYTHON
# ============================================================================

print("\n" + "=" * 60)
print("PART 2: DATA TYPES")
print("=" * 60)
```

```python
# ----------------------------------------------------------------------------
# 2.1 NUMERIC TYPES
# ----------------------------------------------------------------------------

# Integer - whole numbers
int_num = 1
print("Integer:", int_num, type(int_num))

# Float - decimal numbers
float_num = 1.5
print("Float:", float_num, type(float_num))

# Complex - complex numbers
complex_num = complex(3, 5)
print("Complex:", complex_num, type(complex_num))
```

```python
# ----------------------------------------------------------------------------
# 2.2 SEQUENCE TYPES
# ----------------------------------------------------------------------------

# String - sequence of characters
string_var = "Madhav"
string_var2 = '26'
string_var3 = """Multi-line
string example"""
print("String:", string_var, type(string_var))

# List - mutable, ordered, allows duplicates
list_var = [1, 4, 7, 26, 108, 'aditya']
print("List:", list_var, type(list_var))
# Lists can be modified
list_var[0] = 100
print("Modified List:", list_var)

# Tuple - immutable, ordered, allows duplicates
tuple_var = (1, 4, 7, 26, 108, 'aditya')
print("Tuple:", tuple_var, type(tuple_var))
# Tuples CANNOT be modified - uncommenting next line would cause error
# tuple_var[0] = 100  # ERROR!
```
