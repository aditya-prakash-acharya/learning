```python
# 1. Required Arguments (single/multiple aguments)

# ex1 Single arguments
def greetings(name): #name is aparameter
    print("hello", name, "!")

greetings("Aditya") #aditya is argument
#greetings() requied argument to run code

#ex2 multiple argumemts
def intro(course_name, instructor_name):
    print("Welcome to", course_name, "Course by", instructor_name)

intro("python", "Risabh Mishra")
```

```python
# 2. Default Arguments
# we can assign default values tothe fuction if the values 
# not assignedin function it will consider default values
#function defined using one parameter and default value

def greetings(name = "world"):
    print("hello", name, "!")
greetings()
greetings("Aditya")
```

```python
# 3. Keyword Arguments (named arguments)

def divide(a, b): #a and b are two parameters
    return a / b

result = divide(b=10, a=20) # with keyword argument
print(result)

result = divide(10, 20) #positional argument
print(result)
```

```python
# # 4. Arbitrary Arguments ( variable length arguments *args and **kwargs)

# # if we are unsure how many arguments will be passed we will be using *args to accept any number of arguments
# Use this when you don't know how many normal values (numbers, text) you will pass to the function. The * packs them into a tuple (a list you can't change).

#Q1 The Best Example: Calculating Total Price of Items Imagine you are at a shop. You might buy 1 item, or 5 items. You don't know the count beforehand.

def total_price(*items):
    # 'items' TURNS INTO TUPLE LIKE (10, 20, 50)
    # WE ARE USING 'sum()' FOR AUTO ADDING ALL VALUES
    return sum(items)

# Example 1
print(total_price(10, 20, 50))  # Output: 80

# Example 2
print(total_price(5, 5))        # Output: 10

#Q2 Creating a User Profile Imagine you are making a profile. Sometimes you only know the name. Sometimes you know the name, age, and city.

def create_profile(**details):
    print("User Profile Created:")
    
    # check only name
    if 'name' in details:
        print(f"Name: {details['name']}")

# Example: name only
create_profile(name="Alex")

#ex2

def create_profile(**details):
    print("User Profile Created:")
    
    # Check 1: Name
    if 'name' in details:
        print(f"Name: {details['name']}")
    
    # Check 2: Age
    if 'age' in details:
        print(f"Age: {details['age']}")
    
    # Check 3: City
    if 'city' in details:
        print(f"City: {details['city']}")

# Example 1: : Naam, Age, City teeno
create_profile(name="Sam", age=30, city="London")
```
